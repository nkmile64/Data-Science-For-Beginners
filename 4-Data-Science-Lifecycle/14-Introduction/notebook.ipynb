# NYC Taxi data in Winter and Summer

Refer to the [Data dictionary](https://www1.nyc.gov/assets/tlc/downloads/pdf/data_dictionary_trip_records_yellow.pdf) to learn more about the columns that have been provided.


```python
import pandas as pd

path = "../../data/taxi.csv"

# Load the csv file into a dataframe
df = pd.read_csv(path)

# Print the dataframe
df
```

```python
df.info()
```

```python
# Convert the columns that hold datetime values, to Python datetime objects
df["tpep_pickup_datetime"] = pd.to_datetime(df["tpep_pickup_datetime"], format="%Y-%m-%d %H:%M:%S")
df["tpep_dropoff_datetime"] = pd.to_datetime(df["tpep_dropoff_datetime"], format="%Y-%m-%d %H:%M:%S")
```

```python
df.info()
```

```python
# Group rows by dropoff datetime on a monthly basis and calculate average tips per month.
df.groupby([pd.Grouper(key="tpep_dropoff_datetime", freq="ME")])[["tip_amount"]].mean()
```
