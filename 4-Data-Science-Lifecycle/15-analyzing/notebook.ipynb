# Analyzing Data
Examples of the Pandas functions mentioned in the [lesson](README.md).

```python
import pandas as pd
import matplotlib.pyplot as plt

# Loading the dataset
path = "../../data/emails.csv"
email_df = pd.read_csv(path)
```

```python
# Using Describe on the email dataset
email_df.describe()
```

```python
# Sampling 10 emails
email_df.sample(10)
```

```python
# Returns rows where there are more occurrences of "to" than "the"
email_df.query("the < to")
```

```python
# Remove the "Email No." string column, since we want to calculate the
# sum of the rest numeric columns.
email_df.drop(["Email No."], axis=1, inplace=True)
```

```python
# Create a frequency table with the counts of each word in all the emails.
email_word_count = email_df.sum(axis=0).sort_values(ascending=False)
email_word_count
```

From the **frequency table** we can see that "a" is the most common word followed by "i", "on" and "in".

```python
# Plot the frequency table as a histogram
fig, ax = plt.subplots(figsize=(10, 6), layout="constrained")

ax.bar(email_word_count.index, email_word_count.to_numpy(), color="darkcyan")
plt.show()
```
