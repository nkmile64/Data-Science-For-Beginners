# NYC Taxi data in Winter and Summer

Refer to the [Data dictionary](https://www1.nyc.gov/assets/tlc/downloads/pdf/data_dictionary_trip_records_yellow.pdf) to learn more about the columns that have been provided.


```python
import pandas as pd
import numpy as np

path = "../../data/taxi.csv"

# Load the csv file into a dataframe
df = pd.read_csv(path)

# Print the dataframe
df
```

# Exploratory Data Analysis (EDA)

```python
df.info()
```

The dataset contains **200 entries** of **18 columns** each. There are no missing values.<br/>
To answer the question of whether passengers tip drivers more on summer or winter, the columns of interest would be:
- tpep_pickup_datetime
- tpep_dropoff_datetime (actually one date is enough)
- tip_amount
- fare_amount

```python
df = df[["tpep_pickup_datetime", "tip_amount", "fare_amount"]]
df
```

```python
df.info()
```

We must convert the pickup datetime column to a pandas datetime, since it is stored as a plain string in the dataset.

```python
df["tpep_pickup_datetime"] = pd.to_datetime(df["tpep_pickup_datetime"], format="ISO8601")
df.info()
```

Now we want to examine whether there is evidence of any **seasonal tipping behavior**.<br/>
Since the dataset contains records for the months of July and January only, we should aggregate tips for each of these two months and calculate the mean per month.

```python
df.groupby(df["tpep_pickup_datetime"].dt.month)[["tip_amount", "fare_amount"]].mean()
```

It seems that the mean tip amount is slightly higher during the summer compared with winter.
